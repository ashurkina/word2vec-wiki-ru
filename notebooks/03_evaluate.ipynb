## Оценка качества моделей Word2Vec

В этом ноутбуке будет проведена оценка качества обученных моделей Word2Vec. Для этого будет сформирован набор аналогий на основе исходного списка пар слов, после чего модели будут протестированы на задаче решения семантических и синтаксических аналогий. Полученные результаты позволят сравнить качество моделей CBOW и Skip-gram.

## План работы

1. Подготовка набора аналогий.
2. Загрузка обученных моделей.
3. Оценка качества моделей на задаче аналогий.
4. Сравнение результатов CBOW и Skip-gram.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))

from src.analogy_pairs import AnalogyPairsBuilder
from gensim.models import Word2Vec
```

```python
# создаем генератор пар
builder = AnalogyPairsBuilder(
    input_path="../data/tests/analogy-seeds-ru.txt",
    output_path="../data/tests/analogy-dataset-ru.txt",
)

# создаем файл
builder.build()
```

## Загружаем модели

```python
# загружаем модель cbow
cbow_model = Word2Vec.load("../models/cbow_100d.model")

# загружаем модель skip-gram
#skipgram_model = Word2Vec.load("../models/word2vec_skipgram_100d.model")
```

```python
# выводим информацию о моделях
print(f"CBOW: {len(cbow_model.wv)} слов")
#print(f"Skip-gram: {len(skipgram_model.wv)} слов")
```

## Оценка качества

Для оценки качества модели будет использован подготовленный набор аналогий. Метод `evaluate_word_analogies()` вычисляет долю правильно решенных аналогий и возвращает итоговую точность модели.

```python
from src.evaluation import Evaluation
```

```python
# оцениваем модель
evaluation = Evaluation(
    model=cbow_model,
    dataset_path="../data/tests/analogy-dataset-ru.txt",
)

report = evaluation.evaluate()
```

```python
print(f"Accuracy: {report.overall_accuracy:.2%}")
```
