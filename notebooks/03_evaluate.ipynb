## Оценка качества моделей Word2Vec

В этом ноутбуке будет проведена оценка качества обученных моделей Word2Vec. ## Загружаем модели

```python
# загружаем модель cbow
cbow_model = Word2Vec.load("../models/cbow_100d.model")

# загружаем модель skip-gram
skipgram_model = Word2Vec.load("../models/skipgram_100d.model")

# загружаем модель cbow 8e
cbow_8e_model = Word2Vec.load("../models/cbow_100d_8e.model")

# загружаем модель skip-gram 8e
skipgram_8e_model = Word2Vec.load("../models/skipgram_100d_8e.model")

```

## Оценка качества

Для оценки качества модели будет использован подготовленный набор аналогий. Метод `evaluate_word_analogies()` вычисляет долю правильно решенных аналогий и возвращает итоговую точность модели.

```python
from src.evaluation import Evaluation
```

```python
reports = {
    "CBOW (5 эпох)": Evaluation(
        cbow_model,
        "../data/tests/analogy-dataset-ru.txt",
    ).evaluate(),
    "Skip-gram (5 эпох)": Evaluation(
        skipgram_model,
        "../data/tests/analogy-dataset-ru.txt",
    ).evaluate(),
    "CBOW (8 эпох)": Evaluation(
        cbow_8e_model,
        "../data/tests/analogy-dataset-ru.txt",
    ).evaluate(),
    "Skip-gram (8 эпох)": Evaluation(
        skipgram_8e_model,
        "../data/tests/analogy-dataset-ru.txt",
    ).evaluate(),
}
```

```python
rows = []

for name, report in reports.items():
    semantic = report.by_type.loc[
        report.by_type["Тип"] == "Semantic", "Accuracy"
    ].iloc[0]

    syntactic = report.by_type.loc[
        report.by_type["Тип"] == "Syntactic", "Accuracy"
    ].iloc[0]

    rows.append({
        "Модель": name,
        "Общая Accuracy": report.overall_accuracy,
        "Semantic": semantic,
        "Syntactic": syntactic,
    })

results = pd.DataFrame(rows)

for column in ["Общая Accuracy", "Semantic", "Syntactic"]:
    results[column] = (results[column] * 100).round(2)

results
```
