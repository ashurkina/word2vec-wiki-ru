### Обучение моделей Word2Vec

В этом ноутбуке будут обучены модели Word2Vec на подготовленном корпусе русскоязычной Википедии. Будут обучены две архитектуры — CBOW и Skip-gram, после чего модели будут сохранены для дальнейшей оценки качества.

## План работы

1. Загрузка подготовленного корпуса.
2. Обучение модели CBOW.
3. Обучение модели Skip-gram.
4. Сохранение обученных моделей.

```python
from pathlib import Path
import sys
import logging

from gensim.models import Word2Vec

from src.tokenizer import Tokenizer
from src.corpus import Corpus

PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))


logging.basicConfig(
    format="%(asctime)s : %(levelname)s : %(message)s",
    level=logging.INFO,
)
```

```python
corpus_path = Path("../data/processed/corpus_wiki_ru_07_26.txt")
```

```python
tokenizer = Tokenizer()
corpus = Corpus(corpus_path, tokenizer)
```

```python
# обучаем модель cbow
cbow_model = Word2Vec(
    sentences=corpus,
    vector_size=100,
    window=5,
    min_count=50,
    sg=0,
    seed=42,
    workers=8,
    epochs=5,
)
```

```python
MODEL_PATH = Path("../models/cbow_100d.model")

MODEL_PATH.parent.mkdir(
    parents=True,
    exist_ok=True,
)

cbow_model.save(str(MODEL_PATH))

print(f"Model saved to {MODEL_PATH}")
```

```python
# обучаем модель skip-gram
skipgram_model = Word2Vec(
    sentences=corpus,
    vector_size=100,
    window=5,
    min_count=50,
    sg=1,
    seed=42,
    workers=8,
    epochs=5,
)
```

```python
MODEL_PATH = Path("../models/skipgram_100d.model")

MODEL_PATH.parent.mkdir(
    parents=True,
    exist_ok=True,
)

skipgram_model.save(str(MODEL_PATH))

print(f"Model saved to {MODEL_PATH}")
```
